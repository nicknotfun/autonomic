```python
from amp.codec import connect
from amp.system import System
import asyncio

with connect("10.1.0.201", trace=True) as transport:
    with System(transport) as system:
        try:
            await system.discover()
        except asyncio.TimeoutError:
            print("Discovery timed out")

        await asyncio.sleep(1)
        system.dump()
```

```python
from amp.codec import connect
from amp.system import System
import asyncio

transport = connect("10.1.0.201", trace=False, read_only=False)
system = System(transport)
try:
    await asyncio.wait_for(system.discover(), timeout=10)
except asyncio.TimeoutError:
    print("Discovery timed out")

await asyncio.sleep(1)
system.dump()
```
