```python
import asyncio

from amp.system import System

HOSTS = ("10.1.0.200","10.1.0.201")
VOLUME = 0.5
```

```python
system = System(HOSTS, trace=False, read_only=False)

try:
    await system.discover(target_devices=2, timeout_secs=10)
except asyncio.TimeoutError:
    print("Discovery timed out")

system.state.dump()
await system.state.save_to_file("state.json")
```

```python
all_outputs = system.all_outputs()
w1 = system.input_by_name("W1")

print(f"W1 selector: 0x{w1.selector:02X}")
```

```python
# Enable everything.
all_outputs.enable()
```

```python
# Set all outputs to the same volume.
all_outputs.set_volume(VOLUME)
```

```python
# Set all outputs to the input named W1.
all_outputs.set_input(w1)
```

```python
# Mute everything.
all_outputs.mute()
```

```python
system.refresh_outputs()
await asyncio.sleep(2)

# Dump all outputs current inputs
for output in system.all_outputs(expand=True):
    if output.name:
        print(f"{output.name}: {output.input.qualified_name} {output.input.name}")
```
